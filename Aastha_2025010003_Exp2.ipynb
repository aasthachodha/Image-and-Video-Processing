```python
import cv2

img = cv2.imread("image.jpg", 0)

threshold = img.mean()

binary = (img >= threshold).astype('uint8') * 255

cv2.imshow("Original Grayscale", img)
cv2.imshow("Binary Image", binary)
```

```python
import cv2

img = cv2.imread("image.jpg", 0)

threshold = int(input("Enter threshold value (0-255): "))

binary = (img >= threshold).astype('uint8') * 255

cv2.imshow("Original Grayscale", img)
cv2.imshow("Binary Image", binary)
```

```python
import cv2
import numpy as np

img = cv2.imread("image.jpg")

B, G, R = cv2.split(img)

gray = ((R.astype(float) + G.astype(float) + B.astype(float)) / 3).astype(np.uint8)

cv2.imshow("Original RGB Image", img)
cv2.imshow("Grayscale Image", gray)
```

```python
import cv2

img = cv2.imread("image.jpg")
B, G, R = cv2.split(img)

wr = float(input("R weight: "))
wg = float(input("G weight: "))
wb = float(input("B weight: "))

if 0 <= wr <= 1 and 0 <= wg <= 1 and 0 <= wb <= 1 and wr + wg + wb == 1:
    gray = (R*wr + G*wg + B*wb).astype('uint8')

    cv2.imshow("Grayscale", gray)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Invalid weights")
```
