```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

# Image size
rows, cols = 256, 256

# Generate random noise using mathematical equations

# 1. Uniform Noise
uniform = np.random.uniform(0, 1, (rows, cols))

# 2. Gaussian Noise
gaussian = np.random.normal(0, 0.1, (rows, cols))

# 3. Erlang Nois
erlang = np.random.gamma(3, 0.1, (rows, cols))

# 4. Exponential Noise
exponential = np.random.exponential(0.1, (rows, cols))

# 5. Rayleigh Noise
rayleigh = np.random.rayleigh(0.1, (rows, cols))

def display_noise(noise, title, index):

    plt.subplot(5, 2, index * 2 - 1)
    plt.imshow(noise, cmap='gray')
    plt.title(title)
    plt.axis('off')

    plt.subplot(5, 2, index * 2)
    plt.hist(noise.ravel(), bins=50, density=True)
    plt.title(title + ' Histogram')
    plt.figure(figsize=(12, 18))

display_noise(uniform, 'Uniform Noise', 1)
display_noise(gaussian, 'Gaussian Noise', 2)
display_noise(erlang, 'Erlang Noise', 3)
display_noise(exponential, 'Exponential Noise', 4)
display_noise(rayleigh, 'Rayleigh Noise', 5)

plt.tight_layout()
plt.show()
```
