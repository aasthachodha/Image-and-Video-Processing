```python
# read , display , write

import cv2
img = cv2.imread("car.jpg")

cv2.imshow("IMAGE" , img)
cv2.imwrite("car.jpg" , img)
```

```python
# convert image into PNG and TIFF

import cv2

img = cv2.imread("car.jpg")

cv2.imwrite("car.png", img)
cv2.imwrite("car.tiff", img)

print("Image converted successfully.")
```

```python
# resize an image

import cv2

img = cv2.imread("car.jpg")

resized = cv2.resize(img, (400, 300))
cv2.imshow("Resized Image", resized)
```

```python
# convert colored image into gray scale image

import cv2

img = cv2.imread("image.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow("Grayscale Image", gray)
```

```python
# read and display a video
import cv2

video = cv2.VideoCapture("video.mp4")

while True:
    success, frame = video.read()

    if not success:
        break

    cv2.imshow("Video", frame)

    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

video.release()
```

```python
# extract frames from the video

import cv2

video = cv2.VideoCapture("video.mp4")
count = 0

while True:
    success, frame = video.read()

    if not success:
        break

    cv2.imwrite(f"frame{count}.jpg", frame)
    count += 1

video.release()
print("Frames extracted successfully.")
```
