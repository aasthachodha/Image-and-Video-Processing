```python
import cv2

img = cv2.imread("image.jpg")
width = int(input("Enter border width: "))

print("Enter border color in BGR format:")
b = int(input("Blue: "))
g = int(input("Green: "))
r = int(input("Red: "))

color = (b, g, r)
bordered = cv2.copyMakeBorder(
    img,
    width, width, width, width,
    cv2.BORDER_CONSTANT,
    value=color
)
cv2.imshow("Original Image", img)
cv2.imshow("Image with Border", bordered)
```

```python
import cv2

img = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)

width = int(input("Enter border width: "))
color = int(input("Enter border intensity (0-255): "))

bordered = cv2.copyMakeBorder(
    img,
    width, width, width, width,
    cv2.BORDER_CONSTANT,
    value=color
)

cv2.imshow("Original", img)
cv2.imshow("Bordered", bordered)
```

```python
import cv2

img = cv2.imread("binary.jpg", cv2.IMREAD_GRAYSCALE)

# Make sure image is binary
_, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

complement = 255 - binary

cv2.imshow("Binary Image", binary)
cv2.imshow("Complement", complement)
```
